# Named Entity Recognition Application

An NLP application for extracting and categorizing named entities from text.

## 1. Install and Import

```python
!pip -q install spacy
!python -m spacy download en_core_web_sm -q

import spacy
import pandas as pd
import matplotlib.pyplot as plt

nlp = spacy.load("en_core_web_sm")
print("Named Entity Recognition model loaded.")
```

## 2. Entity Extraction

```python
def extract_entities(text):
    doc = nlp(text)
    return pd.DataFrame([
        {"Entity": e.text, "Type": e.label_, "Start": e.start_char, "End": e.end_char}
        for e in doc.ents
    ])

sample = (
    "Elon Musk founded SpaceX in California. "
    "Microsoft announced a new AI project in Seattle in 2025. "
    "Google is headquartered in Mountain View."
)

entities = extract_entities(sample)
display(entities)
```

## 3. Entity Categories

```python
entity_labels = {
    "PERSON": "Person",
    "ORG": "Organization",
    "GPE": "Country / City / State",
    "LOC": "Location",
    "DATE": "Date",
    "MONEY": "Money",
    "PRODUCT": "Product",
    "EVENT": "Event",
    "CARDINAL": "Number"
}

def entity_meaning(label):
    return entity_labels.get(label, label)

if not entities.empty:
    entities["Meaning"] = entities["Type"].apply(entity_meaning)
    display(entities)
```

## 4. Application

```python
def entity_extractor():
    print("=" * 60)
    print("             NAMED ENTITY EXTRACTOR")
    print("=" * 60)
    print("Enter a paragraph. Press ENTER on an empty line to finish.\n")

    lines = []
    while True:
        line = input()
        if line == "":
            break
        lines.append(line)

    text = " ".join(lines)

    if not text.strip():
        print("No text entered.")
        return

    result = extract_entities(text)

    if result.empty:
        print("No named entities detected.")
        return

    result["Meaning"] = result["Type"].apply(entity_meaning)

    print("\nDETECTED ENTITIES")
    print("-" * 60)
    for _, row in result.iterrows():
        print(f"{row['Entity']} -> {row['Type']} ({row['Meaning']})")

# Uncomment to run:
# entity_extractor()
```

## 5. Entity Statistics

```python
if not entities.empty:
    counts = entities["Type"].value_counts()
    display(counts.to_frame("Count"))

    plt.figure(figsize=(8,4))
    counts.plot(kind="bar")
    plt.title("Named Entity Distribution")
    plt.xlabel("Entity Type")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## Conclusion

The application extracts people, organizations, locations, dates, and other named entities from unstructured text.
